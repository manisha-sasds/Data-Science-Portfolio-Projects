```python
import pandas as pd
import numpy as np
from pathlib import Path 


data_path=Path("/Users/manishasasatte/Downloads/eon_volume_forecasting_python_practice")


consumption=pd.read_csv(data_path/"half_hourly_consumption.csv")
weather = pd.read_csv(data_path / "weather.csv")
customers = pd.read_csv(data_path / "customer_master.csv")
future_weather = pd.read_csv(data_path / "future_weather_forecast.csv")
calendar= pd.read_csv("../data/raw/calendar.csv")
print(consumption.head(3))
print(calendar.head(2))
print(future_weather.head(4))


```

```python
def data_overview(df, name):
    print("=" * 60)
    print(f"DATASET: {name}")
    print("=" * 60)

    print("\nShape:")
    print(df.shape)

    print("\nColumns:")
    print(df.columns.tolist())

    print("\nData Types:")
    print(df.dtypes)

    print("\nMissing Values:")
    print(df.isnull().sum())

    print("\nDuplicate Rows:")
    print(df.duplicated().sum())

    print("\nUnique Values:")
    print(df.nunique())

    print("\nFirst 5 Rows:")
    display(df.head())



data_overview(consumption, "Consumption")
data_overview(weather, "Weather")
data_overview(customers, "Customers")
data_overview(future_weather, "Future Weather")
data_overview(calendar, "Calendar")
```

```python
# we storted this sorted_cons df
consumption=consumption.sort_values(["customer_id","timestamp"])
consumption
```

```python
duplicate_readings = consumption[
    consumption.duplicated(
        subset=["customer_id", "timestamp"],
        keep=False
    )
]
duplicate_readings
```

```python
# Step 11 — Check missing values again
consumption.isnull().sum()
negative_consumption = consumption[
    consumption["consumption_kwh"] < 0
]
negative_consumption

zero_consumption = consumption[
    consumption["consumption_kwh"] == 0
]

zero_consumption.head()
```

```python
consumption["time_difference"] =consumption.groupby("customer_id")["timestamp"].diff()
consumption["time_difference"]
```

```python
# Date  features
consumption["date"]=consumption["timestamp"].dt.date
consumption["year"]=consumption["timestamp"].dt.year
consumption["month"]=consumption["timestamp"].dt.month
consumption["day"]=consumption["timestamp"].dt.day
consumption["hour"]=consumption["timestamp"].dt.hour
consumption["minute"]=consumption["timestamp"].dt.minute
consumption["day_of_week"]=consumption["timestamp"].dt.day_of_week
consumption["day_name"]=consumption["timestamp"].dt.day_name()


consumption
```
